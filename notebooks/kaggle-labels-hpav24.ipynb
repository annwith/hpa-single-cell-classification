```python
import pandas as pd
import numpy as np

from collections import Counter
```

```python
# De onde que eu peguei isso mesmo?
kaggle_locations = dict({
    "nucleoplasm": 0,
    "nuclear membrane": 1,
    "nucleoli": 2,
    "nucleoli fibrillar center": 3,
    "nuclear speckles": 4,
    "nuclear bodies": 5,
    "endoplasmic reticulum": 6,
    "golgi apparatus": 7,
    "intermediate filaments": 8,
    "actin filaments": 9,
    "focal adhesion sites": 9,
    "microtubules": 10,
    "mitotic spindle": 11,
    "centrosome": 12,
    "centriolar satellite": 12,
    "plasma membrane": 13,
    "cell junctions": 13,
    "mitochondria": 14,
    "aggresome": 15,
    "cytosol": 16,
    "vesicles": 17,
    "peroxisomes": 17,
    "endosomes": 17,
    "lysosomes": 17,
    "lipid droplets": 17,
    "cytoplasmic bodies": 17,
    "no staining": 18
})
```

```python
df = pd.read_csv("v24_hpa_subcellular_location_expanded.csv")
df = df[['Gene ID', 'Cell Line', 'Location', 'Image']]
df['Image'] = df['Image'].apply(lambda x: x.split('_blue_red_green.jpg')[0])
df['Image'] = df['Image'].apply(lambda x: x.split('/')[-1])

df['Location'] = df['Location'].apply(lambda x: eval(x))
df['Location'] = df['Location'].apply(lambda x: [" ".join(loc.split()[:-1]) for loc in x])
# Colocar "no staining" onde tem []
df['Location'] = df['Location'].apply(lambda x: ["no staining"] if x == [] else x)

print(len(df), "rows of HPA data fetched.")
df.head(10)
```

```python
# Mostrar todas as locations
hpav24_locations = set()
for loc in df['Location']:
    hpav24_locations.update(loc)
hpav24_locations
```

```python
dict_locations = set(kaggle_locations.keys())

only_in_df = hpav24_locations - dict_locations
only_in_df
```

```python
only_in_dict = dict_locations - hpav24_locations
only_in_dict
```

```python
in_both = dict_locations & hpav24_locations
len(in_both), in_both
```

```python
counter = Counter()

for locs in df['Location']:
    for loc in locs:
        if loc in only_in_df:
            counter[loc] += 1

counter
```

```python
from collections import Counter

counter = Counter()

for locs in df['Location']:
    for loc in locs:
        if loc in in_both:
            counter[loc] += 1

counter
```

```python
# Keep only rows where all locations are in in_both
df_filtered = df[df['Location'].apply(lambda labels: set(labels).issubset(in_both))]
len(df_filtered)
```

```python
# Pegar todos os nomes de arquivos dentro de uma pasta ../hpa-datasets/v24/images/cell_crops
import os

# Pasta com os cell crops
cell_crops_dir = "../hpa-datasets/v24/images/cell_crops"

# Listar arquivos
files = [
    f for f in os.listdir(cell_crops_dir)
    if f.endswith(".png")
]

# Extrair ID da imagem (tudo antes do último "_")
image_ids = [
    "_".join(f.split("_")[:-1]) for f in files
]

# Contar número de células por imagem
cell_counts = Counter(image_ids)
```

```python
len(cell_counts)
```

```python
# Verificar se todos os ids de cell_counts (Counter) estão em df_filtered
ids_cell_counts = set(cell_counts.keys())
ids_df = set(df_filtered["Image"])

missing_ids = ids_cell_counts - ids_df
len(missing_ids), list(missing_ids)[:10]
```

```python
locs = df_filtered[df_filtered['Image']=='1027_F12_7']['Location']
len(locs)
```

```python
NUM_CLASSES = 19

rows = []

for image_id, n_cells in cell_counts.items():
    # Pega as localizações dessa imagem
    locs = df_filtered[df_filtered['Image']==image_id]['Location']

    if len(locs) > 0:
        row = {
            "ID": image_id,
            "fold": -1,
            "idx": n_cells
        }
        # Inicializa todas as classes com 0
        for i in range(NUM_CLASSES):
            row[f"class{i}"] = 0
        
        locs = locs.iloc[0]  # lista de strings

        for loc in locs:
            if loc in kaggle_locations:
                cls_idx = kaggle_locations[loc]
                row[f"class{cls_idx}"] = 1
            else:
                # opcional: logar classes fora do mapeamento
                pass

        rows.append(row)

df_out = pd.DataFrame(rows)
```

```python
ordered_cols = (
    ["ID"]
    + [f"class{i}" for i in range(NUM_CLASSES)]
    + ["fold", "idx"]
)

df_out = df_out[ordered_cols]

df_out
```

```python
df_out.to_csv("kaggle_labels_hpa_v24.csv", index=False)
```
